```python
import numpy as np
import matplotlib.pyplot as plt

from analytical.pseudopermanent_R_1d import radial_pseudo_1d

# parameters

Po=25e6                 # pressão inicial [Pa]
qw=100/86400            # vazão [m³/s] = 100 m³/dia
h=20                    # espessura [m]
k=100*9.869233e-16      # permeabilidade [m²] = 100 mD
re=500                  # raio externo [m]
rw=0.1                  # raio do poço [m]
theta=0.20              # porosidade [-]
ct=1e-9                 # compressibilidade total [1/Pa]
u=2e-3                  # viscosidade [Pa.s] = 2 cP
t_final=30*86400        # 30 dias em segundos


r,t,R,T,P=radial_pseudo_1d(
    Po,
    qw,
    h,
    k,
    re,
    rw,
    theta,
    ct,
    u,
    t_final
)

# P em função de r

t_escolhido=10*86400

indice_t=np.argmin(np.abs(t-t_escolhido))

plt.plot(r,P[indice_t,:]/1e6)

plt.xlabel('r [m]')
plt.ylabel('P [MPa]')
plt.title(f'Pressão em função de r - t={t[indice_t]/86400:.1f} dias')

plt.grid()
plt.show()

# P em função de t

r_escolhido=100

indice_r=np.argmin(np.abs(r-r_escolhido))

plt.plot(t/86400,P[:,indice_r]/1e6)

plt.xlabel('t [dias]')
plt.ylabel('P [MPa]')
plt.title(f'Pressão em função de t - r={r[indice_r]:.1f} m')

plt.grid()
plt.show()
```
