```python
import numpy as np
import matplotlib.pyplot as plt

from analytical.pseudopermanent_R1d import radial_pseudo_1d

# parameters

p0 = 35.17e6                 # pressão inicial [Pa]
qw = 1.412e-3         # vazão volumétrica no reservatório [m³/s]
h = 37                   # espessura [m]
k = 1.461e-14       # permeabilidade [m²]  =  14.8 mD
re = 500                  # raio externo [m]
rw = 0.125                  # raio do poço [m]
phi = 0.21               # porosidade [-]
ct = 1.448e-9       # compressibilidade total [1/Pa]
mu = 1.82e-3                 # viscosidade [Pa.s]  =  2 cP
t_final = 45*86400        # 45 dias em segundos


r, t, R, T, P = radial_pseudo_1d(p0, qw, h, k, re, rw, phi, ct, mu,t_final)

# P em função de r

t_escolhido = 20*86400

indice_t = np.argmin(np.abs(t-t_escolhido))

plt.plot(r, P[indice_t,:]/1e6)

plt.xlabel('r [m]')
plt.ylabel('P [MPa]')
plt.title(f'Radial pseudopermanente - P(r) em t = {t[indice_t]/86400:.0f} dias')

plt.grid()
plt.show()

# P em função de t

r_escolhido = 100

indice_r = np.argmin(np.abs(r-r_escolhido))

plt.plot(t/86400,P[:,indice_r]/1e6)

plt.xlabel('t [dias]')
plt.ylabel('P [MPa]')
plt.title(f'Radial Pseudopermanente - P(t) em r = {r[indice_r]:.0f} m')

plt.grid()
plt.show()
```
